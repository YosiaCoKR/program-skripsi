```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("DATASET-BERAS.csv")
```

```python
df.shape
df.info()
df.dtypes
```

```python
df.isnull().sum()
sns.heatmap(df.isnull(), cbar=False)
```

```python
(df.isnull().sum() / len(df)) * 100
```

```python
df["tanggal"] = pd.to_datetime(df["tanggal"])
df.set_index("tanggal", inplace=True)
```

```python
feature_interpolation = df.interpolate(method="time", limit_direction="both", axis=0)
```

```python
df_cleaned = feature_interpolation
print(df_cleaned.isnull().sum())
```

```python
sns.heatmap(df_cleaned.isnull(), cbar=False)
```
